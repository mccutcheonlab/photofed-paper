```python
# Compare grain-fraction ratios to 0.5 for each diet phase
# Primary test: Wilcoxon signed-rank (non-parametric, paired to 0.5)
# Reference test: one-sample t-test
from scipy import stats

diff_points = diff_stats["diff_points"]

diff_test_rows = []
alpha = 0.05
n_phases = len(phases)

for phase in phases:
    vals = diff_points.groupby("dietphase")["diff"].get_group(phase).to_numpy() if phase in diff_points.index.get_level_values("dietphase") else np.array([])
    vals = vals[~np.isnan(vals)]
    n = len(vals)

    if n == 0:
        diff_test_rows.append({
            "dietphase": phase,
            "n": 0,
            "mean_diff": np.nan,
            "median_diff": np.nan,
            "wilcoxon_stat": np.nan,
            "wilcoxon_p": np.nan,
            "ttest_t": np.nan,
            "ttest_p": np.nan,
        })
        continue

    diffs = vals - 0

    # Wilcoxon can fail when all diffs are exactly zero; report NaN in that edge case
    try:
        wres = stats.wilcoxon(diffs, alternative="two-sided", zero_method="wilcox", correction=False)
        w_stat, w_p = float(wres.statistic), float(wres.pvalue)
    except ValueError:
        w_stat, w_p = np.nan, np.nan

    tres = stats.ttest_1samp(vals, popmean=0, nan_policy="omit")
    t_stat, t_p = float(tres.statistic), float(tres.pvalue)

    diff_test_rows.append({
        "dietphase": phase,
        "n": int(n),
        "mean_diff": float(np.mean(vals)),
        "median_diff": float(np.median(vals)),
        "wilcoxon_stat": w_stat,
        "wilcoxon_p": w_p,
        "ttest_t": t_stat,
        "ttest_p": t_p,
    })

diff_tests = pd.DataFrame(diff_test_rows)

# Simple multiplicity control across phases (Bonferroni)
diff_tests["wilcoxon_p_bonf"] = np.minimum(diff_tests["wilcoxon_p"] * n_phases, 1.0)
diff_tests["ttest_p_bonf"] = np.minimum(diff_tests["ttest_p"] * n_phases, 1.0)

diff_tests["wilcoxon_sig_0p05"] = diff_tests["wilcoxon_p_bonf"] < alpha
diff_tests["ttest_sig_0p05"] = diff_tests["ttest_p_bonf"] < alpha

print("Per-phase tests vs 0 (two-sided):")
display(diff_tests)

```

```python
# Two-factor repeated-measures ANOVA on AUC (dietphase x pellettype)
import pingouin as pg

# One AUC value per mouse/phase/pellet is already in auc_df
anova_df = auc_df[["id", "dietphase", "pellettype", "auc"]].copy()

# Keep only mice with complete data across all within-subject conditions
required_cells = anova_df[["dietphase", "pellettype"]].drop_duplicates().shape[0]
complete_ids = (
    anova_df.groupby("id", sort=False)
    .size()
    .loc[lambda s: s == required_cells]
    .index
)
anova_df = anova_df[anova_df["id"].isin(complete_ids)].copy()

print(f"Complete-case mice for RM-ANOVA: {len(complete_ids)}")
print("Cells per mouse required:", required_cells)

rm_aov = pg.rm_anova(
    data=anova_df,
    dv="auc",
    within=["dietphase", "pellettype"],
    subject="id",
    detailed=True,
)

print("Repeated-measures ANOVA (AUC ~ dietphase * pellettype):")
display(rm_aov)

# Optional paired post hocs across the 6 condition combinations
posthoc = pg.pairwise_tests(
    data=anova_df,
    dv="auc",
    within=["dietphase", "pellettype"],
    subject="id",
    padjust="holm",
    parametric=True,
    return_desc=True,
)
print("Pairwise post hoc tests (Holm-corrected):")
display(posthoc)
```

```python
# Two-way ANOVA treating all observations as independent (no repeated-measures structure)
import pingouin as pg

anova_ind_df = auc_df[["id", "dietphase", "pellettype", "auc"]].copy()
anova_ind_df = anova_ind_df.dropna(subset=["auc", "dietphase", "pellettype"])

print("Independent-model sample size summary:")
print("Total observations:", len(anova_ind_df))
print("Unique mice:", anova_ind_df["id"].nunique())
print("Cell counts (dietphase x pellettype):")
display(anova_ind_df.groupby(["dietphase", "pellettype"], sort=False).size().rename("n").reset_index())

# NOTE: This ignores within-mouse dependence and may inflate false positives.
ind_aov = pg.anova(
    data=anova_ind_df,
    dv="auc",
    between=["dietphase", "pellettype"],
    detailed=True,
)
print("Two-way ANOVA (independent model):")
display(ind_aov)

ind_posthoc = pg.pairwise_tests(
    data=anova_ind_df,
    dv="auc",
    between=["dietphase", "pellettype"],
    padjust="holm",
    parametric=True,
    return_desc=True,
)
print("Pairwise post hoc tests (independent model, Holm-corrected):")
display(ind_posthoc)
```
